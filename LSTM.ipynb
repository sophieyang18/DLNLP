```python
import os

def read_txt_files_with_different_encodings(folder_path):
    # 定义常见的编码方式
    encodings = ['utf-8-sig', 'gb18030', 'gbk', 'big5', 'latin1', 'iso-8859-1']
    
    # 初始化存储文件内容和文件名的列表
    file_contents = []
    file_names = []
    
    # 遍历指定文件夹中的所有文件
    for filename in os.listdir(folder_path):
        if filename.endswith(".txt"):
            file_path = os.path.join(folder_path, filename)
            # 尝试不同的编码方式读取文件
            for encoding in encodings:
                try:
                    with open(file_path, 'rb') as file:
                        content = file.read().decode(encoding)
                    # 如果成功读取，则存储内容并跳出循环
                    file_contents.append(content)
                    file_names.append(filename[:-4])  # 去掉.txt后缀
                    break
                except UnicodeDecodeError:
                    continue
            else:
                # 如果所有编码方式都失败，打印错误信息
                print(f"无法解码文件: {filename}")
    
    return file_contents, file_names

# 示例使用
folder_path = r'C:\Users\ASUS\Desktop\研一（下）\dlnlp\第三次大作业\jyxstxtqj_downcc.com'
contents, names = read_txt_files_with_different_encodings(folder_path)

for items in contents:
    print("文件内容:", items[:500])
print("文件标题:", names)
```

```python
import re
from typing import List

def split_into_paragraphs(text: str, min_length: int = 300, max_length: int = 500) -> List[str]:
    """
    将文本分割成300-500字的段落，尽量保持段落完整性
    """
    paragraphs = []
    # 首先按自然段落分割
    natural_paras = re.split(r'\n+', text.strip())
    
    current_para = ""
    for para in natural_paras:
        para = para.strip()
        if not para:
            continue
            
        # 如果当前段落加上新段落不超过最大长度，就合并
        if len(current_para) + len(para) <= max_length:
            current_para += "\n" + para if current_para else para
        else:
            # 如果当前段落已经达到最小长度，就作为一个训练样本
            if len(current_para) >= min_length:
                paragraphs.append(current_para)
                current_para = para
            else:
                # 否则继续累积
                current_para += "\n" + para
    
    # 添加最后一个段落
    if len(current_para) >= min_length:
        paragraphs.append(current_para)
    
    return paragraphs

# 示例使用
# novel_texts = [...]  # 包含所有金庸小说文本的列表
# all_paragraphs = []
# for text in novel_texts:
#     all_paragraphs.extend(split_into_paragraphs(text))
```

```python
from gensim.models import FastText
import jieba

def train_fasttext_model(paragraphs: List[str], model_path: str = "fasttext_jinyong.model"):
    """
    训练FastText词向量模型
    """
    # 分词处理
    sentences = []
    for para in paragraphs:
        words = list(jieba.cut(para))
        sentences.append(words)
    
    # 训练FastText模型
    model = FastText(vector_size=300, window=5, min_count=3, workers=4)
    model.build_vocab(sentences)
    model.train(sentences, total_examples=model.corpus_count, epochs=10)
    
    # 保存模型
    model.save(model_path)
    return model

# 示例使用
# fasttext_model = train_fasttext_model(all_paragraphs)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Embedding, Dropout
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.utils import to_categorical
import numpy as np

class TextGenerator:
    def __init__(self, fasttext_model, seq_length=20):
        self.fasttext_model = fasttext_model
        self.seq_length = seq_length
        self.word_index = {word: i for i, word in enumerate(fasttext_model.wv.index_to_key)}
        self.index_word = {i: word for i, word in enumerate(fasttext_model.wv.index_to_key)}
        self.vocab_size = len(self.word_index)
        self.embedding_matrix = fasttext_model.wv.vectors
        self.model = self._build_model()
    
    def _build_model(self):
        model = Sequential([
            Embedding(input_dim=self.vocab_size,
                      output_dim=300,
                      weights=[self.embedding_matrix],
                      input_length=self.seq_length,
                      trainable=False),
            LSTM(256, return_sequences=True),
            Dropout(0.2),
            LSTM(256),
            Dropout(0.2),
            Dense(self.vocab_size, activation='softmax')
        ])
        
        model.compile(loss='categorical_crossentropy', optimizer='adam')
        return model
    
    def prepare_data(self, paragraphs):
        X, y = [], []
        
        for para in paragraphs:
            words = list(jieba.cut(para))
            for i in range(len(words) - self.seq_length):
                seq = words[i:i + self.seq_length]
                next_word = words[i + self.seq_length]
                
                # 转换为索引
                seq_idx = [self.word_index.get(word, 0) for word in seq]
                next_idx = self.word_index.get(next_word, 0)
                
                X.append(seq_idx)
                y.append(next_idx)
        
        X = pad_sequences(X, maxlen=self.seq_length)
        y = to_categorical(y, num_classes=self.vocab_size)
        
        return X, y
    
    def train(self, X, y, epochs=20, batch_size=128):
        self.model.fit(X, y, epochs=epochs, batch_size=batch_size)
    
    def generate_text(self, seed_text, num_words=100, temperature=1.0):
        generated = []
        input_seq = seed_text
        
        for _ in range(num_words):
            # 分词并获取最后seq_length个词
            words = list(jieba.cut(input_seq))
            if len(words) > self.seq_length:
                words = words[-self.seq_length:]
            elif len(words) < self.seq_length:
                # 填充
                words = [''] * (self.seq_length - len(words)) + words
            
            # 转换为索引
            seq_idx = [self.word_index.get(word, 0) for word in words]
            seq_idx = pad_sequences([seq_idx], maxlen=self.seq_length)
            
            # 预测下一个词
            preds = self.model.predict(seq_idx, verbose=0)[0]
            
            # 应用温度参数
            preds = np.log(preds) / temperature
            exp_preds = np.exp(preds)
            preds = exp_preds / np.sum(exp_preds)
            
            # 采样
            next_idx = np.random.choice(range(self.vocab_size), p=preds)
            next_word = self.index_word.get(next_idx, '')
            
            generated.append(next_word)
            input_seq += next_word
        
        return ''.join(generated)
```

```python
class StreamingTextGenerator:
    def __init__(self, fasttext_model, seq_length=10, max_vocab_size=5000):
        self.fasttext_model = fasttext_model
        self.seq_length = seq_length
        self.max_vocab_size = max_vocab_size
        self._build_vocab()
        self.model = self._build_model()
    
    def _build_vocab(self):
        """构建限定大小的词汇表"""
        # 按词频排序取前max_vocab_size个词
        vocab_items = sorted(self.fasttext_model.wv.key_to_index.items(), 
                           key=lambda x: x[1])[:self.max_vocab_size]
        self.word_index = {word: i+1 for i, (word, _) in enumerate(vocab_items)}  # 0保留给未知词
        self.index_word = {i+1: word for i, (word, _) in enumerate(vocab_items)}
        self.vocab_size = len(self.word_index) + 1  # 包括未知词
        self.embedding_dim = self.fasttext_model.wv.vector_size
        
        # 构建嵌入矩阵 (索引0是全零的未知词向量)
        self.embedding_matrix = np.zeros((self.vocab_size, self.embedding_dim))
        for word, i in self.word_index.items():
            self.embedding_matrix[i] = self.fasttext_model.wv[word]
    
    def _build_model(self):
        """构建LSTM模型"""
        model = Sequential([
            Embedding(input_dim=self.vocab_size,
                    output_dim=self.embedding_dim,
                    weights=[self.embedding_matrix],
                    input_length=self.seq_length,
                    trainable=False,
                    mask_zero=True),
            LSTM(64, return_sequences=True),
            Dropout(0.2),
            LSTM(64),
            Dropout(0.2),
            Dense(self.vocab_size, activation='softmax')
        ])
        model.compile(loss='sparse_categorical_crossentropy', optimizer='adam')
        return model
    
    def data_generator(self, paragraphs, batch_size=32):
        """生成训练批数据的生成器"""
        while True:
            for para in paragraphs:
                words = list(jieba.cut(para))
                X = []
                y = []
                
                for i in range(len(words) - self.seq_length):
                    seq = words[i:i + self.seq_length]
                    next_word = words[i + self.seq_length]
                    
                    seq_idx = [self.word_index.get(word, 0) for word in seq]
                    next_idx = self.word_index.get(next_word, 0)
                    
                    X.append(seq_idx)
                    y.append(next_idx)
                    
                    if len(X) == batch_size:
                        X_padded = pad_sequences(X, maxlen=self.seq_length)
                        yield X_padded, np.array(y)
                        X, y = [], []
            
            # 处理剩余的不足一个batch的数据
            if X:
                X_padded = pad_sequences(X, maxlen=self.seq_length)
                yield X_padded, np.array(y)
    
    def generate_text(self, seed_text, num_words=100, temperature=0.7):
        """改进版生成方法：强制保持开头一致性"""
        generated = list(jieba.cut(seed_text))  # 将种子文本分词并作为初始输出
        input_seq = seed_text
        
        for _ in range(num_words):
            # 始终从原始输入开始构建序列
            context = list(jieba.cut(input_seq))[-self.seq_length:]
            if len(context) < self.seq_length:
                context = [''] * (self.seq_length - len(context)) + context
            
            seq_idx = [self.word_index.get(word, 0) for word in context]
            seq_idx = pad_sequences([seq_idx], maxlen=self.seq_length)
            
            preds = self.model.predict(seq_idx, verbose=0)[0]
            preds = np.log(preds) / temperature
            exp_preds = np.exp(preds)
            preds = exp_preds / np.sum(exp_preds)
            
            next_idx = np.random.choice(range(self.vocab_size), p=preds)
            next_word = self.index_word.get(next_idx, '')
            
            generated.append(next_word)
            input_seq += next_word  # 逐渐扩展输入序列
        
        # 确保开头与种子文本完全一致
        return seed_text + ''.join(generated[len(list(jieba.cut(seed_text))):])
```

```python
def train_in_batches(generator, paragraphs, epochs=1, batch_size=16, steps_per_epoch=1000):
    # 计算每个epoch的实际步数
    total_samples = sum(len(list(jieba.cut(p))) - generator.seq_length for p in paragraphs)
    steps_per_epoch = min(steps_per_epoch, total_samples // batch_size)
    
    # 创建数据生成器
    train_gen = generator.data_generator(paragraphs, batch_size=batch_size)
    
    # 分段训练
    for epoch in range(epochs):
        print(f"Epoch {epoch+1}/{epochs}")
        history = generator.model.fit(
            train_gen,
            steps_per_epoch=steps_per_epoch,
            epochs=1,
            verbose=1
        )
        
        # 每epoch保存检查点
        generator.model.save(f"jinyong_generator_epoch{epoch+1}.h5")
        
    return history
```

```python
import matplotlib.pyplot as plt
def plot_training_loss(loss_values):
    """绘制训练loss曲线"""
    plt.figure(figsize=(10, 6))
    plt.plot(loss_values, label='Training Loss')
    plt.title('Training Loss Over Batches')
    plt.xlabel('Batch Number')
    plt.ylabel('Loss')
    plt.legend()
    
    # 平滑曲线（可选）
    window_size = max(1, len(loss_values) // 100)  # 动态窗口大小
    smooth_loss = np.convolve(loss_values, np.ones(window_size)/window_size, mode='valid')
    plt.plot(smooth_loss, label=f'Smoothed (window={window_size})', alpha=0.7)
    
    plt.grid(True)
    plt.savefig('training_loss.png')  # 保存图像
    plt.show()
```
